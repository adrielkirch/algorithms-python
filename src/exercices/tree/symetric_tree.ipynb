# 101. Symmetric Tree

Given the `root` of a binary tree, *check whether it is a mirror of itself* (i.e., symmetric around its center).

## Example 1:
```
Input: root = [1,2,2,3,4,4,3]
Output: true
```
Tree visualization:
```
       1
      / \
     2   2
    / \ / \
   3  4 4  3
```

## Example 2:
```
Input: root = [1,2,2,null,3,null,3]
Output: false
```
Tree visualization:
```
       1
      / \
     2   2
      \   \
       3   3
```

## Solution: DFS (Recursive Approach)

````python
from typing import Optional

# Definition for a binary tree node
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    """
    DFS Recursive Approach:
    - Compare two nodes recursively
    - Left subtree of root must mirror right subtree of root
    - Time Complexity: O(n) where n is number of nodes
    - Space Complexity: O(h) where h is height (recursion stack)
    """
    
    def isSymmetric(self, root: Optional[TreeNode]) -> bool:
        def dfs(left: Optional[TreeNode], right: Optional[TreeNode]) -> bool:
            # Both nodes are None - symmetric
            if not left and not right:
                return True
            
            # One node is None or values don't match - not symmetric
            if not left or not right or left.val != right.val:
                return False
            
            # Check if left subtree mirrors right subtree
            # Left child of left node should mirror right child of right node
            # Right child of left node should mirror left child of right node
            return (dfs(left.left, right.right) and 
                    dfs(left.right, right.left))
        
        if not root:
            return True
        
        return dfs(root.left, root.right)

# Test cases
# Example 1: [1,2,2,3,4,4,3]
root1 = TreeNode(1)
root1.left = TreeNode(2)
root1.right = TreeNode(2)
root1.left.left = TreeNode(3)
root1.left.right = TreeNode(4)
root1.right.left = TreeNode(4)
root1.right.right = TreeNode(3)

print("Example 1 (Expected True):", Solution().isSymmetric(root1))

# Example 2: [1,2,2,null,3,null,3]
root2 = TreeNode(1)
root2.left = TreeNode(2)
root2.right = TreeNode(2)
root2.left.right = TreeNode(3)
root2.right.right = TreeNode(3)

print("Example 2 (Expected False):", Solution().isSymmetric(root2))
````

## Solution: BFS (Iterative Approach)

````python
from collections import deque

class SolutionBFS:
    """
    BFS Iterative Approach:
    - Use a queue to compare pairs of nodes level by level
    - Each pair should be mirror images of each other
    - Time Complexity: O(n) where n is number of nodes
    - Space Complexity: O(w) where w is maximum width of tree
    """
    
    def isSymmetric(self, root: Optional[TreeNode]) -> bool:
        if not root:
            return True
        
        # Queue stores pairs of nodes to compare
        queue = deque([(root.left, root.right)])
        
        while queue:
            left, right = queue.popleft()
            
            # Both nodes are None - continue
            if not left and not right:
                continue
            
            # One node is None or values don't match - not symmetric
            if not left or not right or left.val != right.val:
                return False
            
            # Add pairs to compare in next iterations
            # Left child of left node pairs with right child of right node
            queue.append((left.left, right.right))
            # Right child of left node pairs with left child of right node
            queue.append((left.right, right.left))
        
        return True

# Test with same examples
print("\n--- BFS Solution ---")
print("Example 1 (Expected True):", SolutionBFS().isSymmetric(root1))
print("Example 2 (Expected False):", SolutionBFS().isSymmetric(root2))
````
