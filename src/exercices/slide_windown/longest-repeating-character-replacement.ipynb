# 424. Longest Repeating Character Replacement

You are given a string s and an integer k. You can choose any character of the string and change it to any other uppercase English character. You can perform this operation at most k times.

Return the length of the longest substring containing the same letter you can get after performing the above operations.

## Example 1:

Input: s = "ABAB", k = 2
Output: 4
Explanation: Replace the two 'A's with two 'B's or vice versa.

## Example 2:

Input: s = "AABABBA", k = 1

Output: 4
Explanation: Replace the one 'A' in the middle with 'B' and form "AABBBBA".
The substring "BBBB" has the longest repeating letters, which is 4.
There may exists other ways to achieve this answer too.

![ChatGPT Image 5 de mai. de 2026, 14_47_20.png](<attachment:ChatGPT Image 5 de mai. de 2026, 14_47_20.png>)

```python
class Solution:
    def characterReplacement(self, s: str, k: int) -> int:
        char_counter = {}
        left = max_substring_size = max_freq = 0

        for right in range(len(s)):
            char = s[right]
            char_counter[char] = char_counter.get(char,0) + 1
            max_freq = max(max_freq , char_counter[char])
            while (right - left + 1) - max_freq > k:
                left_idx = s[left]
                char_counter[left_idx] -= 1
                if char_counter[left_idx] == 0:
                    del char_counter[left_idx] 
                left += 1
            max_substring_size = max(max_substring_size, right - left + 1)

        return max_substring_size



sol = Solution()
print(sol.characterReplacement("ABAB", 2))        # 4
print(sol.characterReplacement("AABABBA", 1))     # 4
print(sol.characterReplacement("AAAA", 2))        # 4
print(sol.characterReplacement("ABCDE", 1))       # 2
print(sol.characterReplacement("BAAAAB", 2))      # 6
print(sol.characterReplacement("AABACCA", 2))     # 5
print(sol.characterReplacement("ABBBCCDAA", 2))   # 5
```
