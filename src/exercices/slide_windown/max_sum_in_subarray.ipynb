# 53. Maximum Subarray


Given an integer array nums, find the

with the largest sum, and return its sum.


Example 1:

Input: nums = [-2,1,-3,4,-1,2,1,-5,4]
Output: 6
Explanation: The subarray [4,-1,2,1] has the largest sum 6.

Example 2:

Input: nums = [1]
Output: 1
Explanation: The subarray [1] has the largest sum 1.

Example 3:

Input: nums = [5,4,-1,7,8]
Output: 23
Explanation: The subarray [5,4,-1,7,8] has the largest sum 23.

 
Constraints:

    1 <= nums.length <= 105
    -10^4 <= nums[i] <= 104


```python
class Solution:
    def maxSubArray(self, nums):
        current_sum = max_sum = nums[0]
        
        for num in nums[1:]:
            current_sum = max(num, current_sum + num)
            max_sum = max(max_sum, current_sum)
        
        return max_sum
    
print(Solution().maxSubArray([-2,1,-3,4,-1,2,1,-5,4]))
```

```python
class Solution:
    def maxSubArray_dp(self, nums):
        """
        DP Version: dp[i] = máximo sum de subarray terminando em posição i
        Time: O(n), Space: O(n)
        """
        n = len(nums)
        
        # dp[i] representa o máximo sum de subarray que termina em i
        dp = [0] * n
        dp[0] = nums[0]
        
        max_sum = dp[0]
        
        for i in range(1, n):
            # Decidir: continuar o subarray anterior ou começar um novo
            dp[i] = max(nums[i], dp[i-1] + nums[i])
            max_sum = max(max_sum, dp[i])
        
        return max_sum


# Test Cases
sol = Solution()

# Test 1: [-2,1,-3,4,-1,2,1,-5,4] -> 6 (subarray [4,-1,2,1])
test1 = [-2,1,-3,4,-1,2,1,-5,4]
result1_dp = sol.maxSubArray_dp(test1)
assert result1_dp == 6, f"Test 1 failed: DP={result1_dp}"
print(f"✓ Test 1: {test1} -> {result1_dp}")

# Test 2: [1] -> 1
test2 = [1]
result2_dp = sol.maxSubArray_dp(test2)
assert result2_dp == 1, f"Test 2 failed: {result2_dp}"
print(f"✓ Test 2: {test2} -> {result2_dp}")

# Test 3: [5,4,-1,7,8] -> 23 (whole array)
test3 = [5,4,-1,7,8]
result3_dp = sol.maxSubArray_dp(test3)
assert result3_dp == 23, f"Test 3 failed: {result3_dp}"
print(f"✓ Test 3: {test3} -> {result3_dp}")

# Test 4: [-3,-2,-1] -> -1 (single element)
test4 = [-3,-2,-1]
result4_dp = sol.maxSubArray_dp(test4)
assert result4_dp == -1, f"Test 4 failed: {result4_dp}"
print(f"✓ Test 4: {test4} -> {result4_dp}")

# Test 5: [0,-1,0] -> 0
test5 = [0,-1,0]
result5_dp = sol.maxSubArray_dp(test5)
assert result5_dp == 0, f"Test 5 failed: {result5_dp}"
print(f"✓ Test 5: {test5} -> {result5_dp}")

print("\n✅ All DP tests passed!")
print("\n📊 Comparison:")
print(f"  Greedy (Kadane): O(1) space, O(n) time")
print(f"  DP: O(n) space, O(n) time")
print(f"  Both produce the same correct results!")
```
