## 45. Jump Game II

Given an integer array nums, you are initially positioned at the array's first index, and each element in the array represents your maximum jump length at that position.

Return the minimum number of jumps to reach the last index.

Example 1:

Input: nums = [2,3,1,1,4]
Output: 2
Explanation: Jump 1 step from index 0 to 1, then 3 steps to the last index.

Example 2:

Input: nums = [2,3,0,1,4]
Output: 2


```
i=0: farthest = max(0, 0+2) = 2
     i==current_end (0==0) → jumps=1, current_end=2
          └─ "explorei tudo do nível 0, próximo nível vai até índice 2"

i=1: farthest = max(2, 1+3) = 4
     i!=current_end (1!=2) → só atualiza farthest

i=2: farthest = max(4, 2+1) = 4
     i==current_end (2==2) → jumps=2, current_end=4
          └─ "explorei tudo do nível 1 (índices 1 e 2), próximo nível vai até índice 4"

loop termina (range(n-1) = range(4), para em i=3)
→ return 2
```

````python
def jump(nums):
    """Retorna o número mínimo de saltos para chegar ao final (greedy com max reach)."""
    n = len(nums)
    if n <= 1:
        return 0
    
    jumps = 0
    current_end = 0  # O fim do alcance atual
    farthest = 0     # O ponto mais longe que podemos alcançar
    # Maximização nivel a nivel, se i chegou no limite == current_end precisamos incrementar o pulo
    for i in range(n - 1):
        # Atualiza o ponto mais longe que podemos alcançar
        farthest = max(farthest, i + nums[i])
        
        # Se alcançamos o fim do alcance atual, temos que fazer outro salto
        if i == current_end:
            jumps += 1
            current_end = farthest
    
    return jumps

# Exemplos de teste
print(jump([2,3,1,1,4]))    # 2
print(jump([2,3,0,1,4]))    # 2
print(jump([0]))            # 0
print(jump([1,2]))          # 1
````

````python
def jumpDP(nums):
    """Solução com DP - min_jumps[i] = mínimo de saltos para chegar em i."""
    n = len(nums)
    if n <= 1:
        return 0
    
    # min_jumps[i] = mínimo de saltos para alcançar o índice i
    min_jumps = [float('inf')] * n
    min_jumps[0] = 0
    
    for i in range(n):
        # Percorre todos os índices que podemos alcançar a partir de i
        for j in range(i + 1, min(i + nums[i] + 1, n)):
            min_jumps[j] = min(min_jumps[j], min_jumps[i] + 1)
    
    return min_jumps[-1]

# Exemplos de teste
print(jumpDP([2,3,1,1,4]))    # 2
print(jumpDP([2,3,0,1,4]))    # 2
print(jumpDP([0]))            # 0
print(jumpDP([1,2]))          # 1
````
