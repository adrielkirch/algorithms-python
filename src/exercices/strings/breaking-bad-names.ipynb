Given a string **fullName** and an array of strings **elements** representing chemical symbols, find the first occurrence of a chemical symbol within the first word of the name and wrap it in square brackets [].

Constraints & Rules:

You must prioritize longer symbols (2 letters) over shorter ones (1 letter) if both match at the same position.

The matching is case-insensitive, but the output must preserve the original casing of the name, except for the brackets.

Only the first valid match found at the beginning of the first word should be processed.

If no match is found, return the original string.

**Example 1:**

Input: fullName = "Luna Night", elements = ["L", "Lu", "N", "Ni"]

Output: "[Lu]na [Ni]ght"

**Exemple 2:**

Input: fullName = "Luna Night", elements = ["L", "Lu", "N", "Ni","H","He","F"]

Output: "[He]lio F [Ni]kel"

```python
from typing import List

def breakingBad(fullName:str, elements:List[str]):    
    names = [single_name.capitalize() for single_name in fullName.split(" ")]
    set_of_elements  = set(elements)
    result = ''
    for singleName in names:
        if singleName[:2] in set_of_elements:
            result += f'[{singleName[:2]}]{singleName[2:]} '
        elif singleName[:1] in set_of_elements:
            result += f'[{singleName[:1]}]{singleName[1:]} '
        else:
    
            result += f'{singleName.capitalize()} '     
    return result.rstrip()

print(breakingBad(fullName="Luna F Night",elements = ["L", "Lu", "N", "Ni","F"]))
```
