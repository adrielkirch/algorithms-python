# 🏰 Problem: Gridland Provinces

The Kingdom of Gridland contains **P** provinces. Each province is defined as a $2 \times N$ grid where each cell in the grid represents a city. Every cell in the grid contains a single lowercase character denoting the first character of the city name corresponding to that cell.

From a city with the coordinates $(r, c)$, it is possible to move to any of the following cells in **1 unit of time** (provided that the destination cell is within the confines of the grid):
*   $(r, c \pm 1)$
*   $(r \pm 1, c)$

A knight wants to visit all the cities in Gridland. He can start his journey in any city and immediately stops his journey after having visited each city **at least once**. Moreover, he always plans his journey in such a way that the total time required to complete it is **minimum**.

After completing his tour of each province, the knight forms a string by concatenating the characters of all the cells in his path. **How many distinct strings can he form in each province?**

---

### 📥 Input Format
*   The first line contains a single integer, **P**, denoting the number of provinces.
*   The subsequent lines describe each province over the following three lines:
    1.  An integer, **N**, denoting the number of columns in the province.
    2.  A string, **S1**, of length **N** (first row).
    3.  A string, **S2**, of length **N** (second row).

### 📤 Output Format
For each province, print the number of distinct strings the knight can form on a new line.

---

### 💡 Example

**Sample Input:**
```text
3
1
a
a
3
dab
abd
5
ababa
babab
```

**Sample Output:**
```text
1
8
2
```

### 📝 Explanation

*   **Province 0 ($N=1$):**
    The knight can only form one string (`aa`).
*   **Province 1 ($N=3$):**
    The knight can form 8 different strings: `abdbad`, `adabdb`, `badabd`, `bdbada`, `dababd`, `dabdba`, `dbabad`, and `dbadab`.
*   **Province 2 ($N=5$):**
    The knight can form 2 different strings: `ababababab` and `bababababa`.

````python
class Solution:
    def __init__(self):
        self.results = set()
        self.grid = []
        self.n = 0
        self.target_len = 0

    def gridlandProvinces(self, n, s1, s2):
        self.n = n
        self.grid = [s1, s2]
        self.target_len = 2 * n
        self.results = set()

        for r in range(2):
            for c in range(n):
                visited = [[False for _ in range(n)] for _ in range(2)]
                self._backtrack(r, c, "", visited)
        
        return len(self.results)

    def _backtrack(self, r, c, current_str, visited, count):
        current_str += self.grid[r][c]
        visited[r][c] = True
        count += 1
        if count == self.target_len:
            self.results.add(current_str)
        else:
            for dr, dc in [(0, 1), (0, -1), (1, 0), (-1, 0)]:
                nr, nc = r + dr, c + dc

                if 0 <= nr < 2 and 0 <= nc < self.n and not visited[nr][nc]:

                    # Checagem de isolamento:
                    # Se eu me mover para (nr, nc), a célula oposta (1-r, c) 
                    # ficará sem saída?
                    opp_r, opp_c = 1 - r, c
                    if not visited[opp_r][opp_c]:
                        # Se a célula oposta só tem UM vizinho disponível (que é o atual),
                        # e nós estamos saindo dela, ela ficará isolada.
                        if self._is_isolated(opp_r, opp_c, visited, (nr, nc)):
                            continue 
                        
                    self._backtrack(nr, nc, current_str, visited, count)
        visited[r][c] = False

    def _is_isolated(self, r, c, visited, next_move):
        # Conta quantos vizinhos a célula (r, c) ainda tem
        available_neighbors = 0
        for dr, dc in [(0, 1), (0, -1), (1, 0), (-1, 0)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 2 and 0 <= nc < self.n and not visited[nr][nc]:
                if (nr, nc) != next_move: # Se o vizinho não for para onde estamos indo
                    available_neighbors += 1
        return available_neighbors == 0
````
